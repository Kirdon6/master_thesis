```python
import torch
import matplotlib.pyplot as plt

class GradualTransition:
    def __init__(self, num_atom_types=4, num_timesteps=1000):
        self.num_atom_types = num_atom_types
        self.num_timesteps = num_timesteps
        
    def get_transition_matrix(self, t):
        """
        At early timesteps (t close to 0): atoms mostly stay the same
        At late timesteps (t close to num_timesteps): atoms switch randomly
        """
        # Progress from 0 to 1 as t increases
        noise_level = t / self.num_timesteps
        
        # Identity matrix (no change)
        identity = torch.eye(self.num_atom_types)
        
        # Uniform matrix (equal probability for all types)
        uniform = torch.ones(self.num_atom_types, self.num_atom_types) / self.num_atom_types
        
        # Interpolate between identity and uniform
        Q_t = (1 - noise_level) * identity + noise_level * uniform
        
        return Q_t
    
    def visualize_transition_probability(self):
        """Show how the probability of staying the same type changes over time"""
        timesteps = [0, 250, 500, 750, 999]
        
        for t in timesteps:
            Q_t = self.get_transition_matrix(t)
            stay_same_prob = Q_t[0, 0]  # Probability of H staying H
            
            print(f"Timestep {t}:")
            print(f"  Probability of staying same type: {stay_same_prob:.3f}")
            print(f"  Probability of switching: {1 - stay_same_prob:.3f}")
            print()

# Example
transition = GradualTransition()
transition.visualize_transition_probability()
```
