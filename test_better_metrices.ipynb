```python
# Simulate some data
import torch
from nano_evaluator import BatchedNanomaterialEvaluator, quick_batch_metric
# Example with batch of structures
batch_size = 64
n_atoms = 100

# Simulate batch data
pred_batch = torch.randn(batch_size, n_atoms, 3, device='cuda')
true_batch = torch.randn(batch_size, n_atoms, 3, device='cuda')

# Full evaluation
evaluator = BatchedNanomaterialEvaluator()
results = evaluator.batched_comprehensive_evaluation(pred_batch, true_batch)

print("Batch Evaluation Results:")
print(f"Mean aligned distance across batch: {results['aligned_assignment']['batch_mean']:.4f}")
print(f"Std of aligned distances: {results['aligned_assignment']['batch_std']:.4f}")
print(f"Best structure in batch: {min(results['aligned_assignment']['mean_distances']):.4f}")
print(f"Worst structure in batch: {max(results['aligned_assignment']['mean_distances']):.4f}")
print(f"Position MAE: {results['position_mae']['batch_mean']:.4f}")

# Quick single metric
quick_metric = quick_batch_metric(pred_batch, true_batch)
print(f"Quick batch metric: {quick_metric:.4f}")
```

```python
# Evaluate a batch with this approach
from nano_evaluator import BatchedNanomaterialEvaluator
batch_size = 16
n_atoms = 50
n_types = 5  # e.g., C, O, N, H, S

# Simulate batch data with types
pred_batch = torch.randn(batch_size, n_atoms, 3, device='cuda')
true_batch = torch.randn(batch_size, n_atoms, 3, device='cuda')
pred_types = torch.randint(0, n_types, (batch_size, n_atoms), device='cuda')
true_types = torch.randint(0, n_types, (batch_size, n_atoms), device='cuda')

# Evaluation with types
evaluator = BatchedNanomaterialEvaluator()
results = evaluator.comprehensive_evaluation_with_types(pred_batch, true_batch, pred_types, true_types)

print("Evaluation Results with Atom Types:")
print(f"Type-constrained aligned distance: {results['type_constrained_aligned']['batch_mean']:.4f}")
print(f"Type accuracy: {results['type_constrained_aligned']['type_metrics']['batch_mean_type_accuracy']:.4f}")
print(f"Type-ignored aligned distance: {results['type_ignored_aligned']['batch_mean']:.4f}")
print(f"Improvement from type constraints: {results['type_ignored_aligned']['batch_mean'] - results['type_constrained_aligned']['batch_mean']:.4f}")
```
